```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
```

```python
X = np.array([[1], [2], [3], [4], [5], [6]])   # X is 2D
y = np.array([2, 4, 6, 8, 10, 12])             # y is 1D
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

model = LinearRegression()        # hyperparameters are set here, in the constructor
model.fit(X_train, y_train)       # learns from data

print("coef_      :", model.coef_)       # trailing _ = learned after fit
print("intercept_ :", model.intercept_)
```

```python
y_pred = model.predict(X_test)
print("predictions:", y_pred)
print("actual     :", y_test)
print("R2 score   :", model.score(X_test, y_test))
```

```python
Xi, yi = load_iris(return_X_y=True)
Xi_train, Xi_test, yi_train, yi_test = train_test_split(Xi, yi, test_size=0.2, random_state=42)

clf = LogisticRegression(max_iter=200)
clf.fit(Xi_train, yi_train)

print("labels       :", clf.predict(Xi_test[:3]))
print("probabilities:\n", clf.predict_proba(Xi_test[:3]))
print("accuracy     :", clf.score(Xi_test, yi_test))
```

```python
sc = StandardScaler()
sc.fit(X_train)                           # learns mean and std from train
print("mean_  :", sc.mean_)
print("scale_ :", sc.scale_)

X_train_s = sc.transform(X_train)         # apply
X_test_s = sc.transform(X_test)           # transform only, never fit on test
print("train scaled:", X_train_s.ravel())
print("test scaled :", X_test_s.ravel())

# shortcut for train only
print("fit_transform:", StandardScaler().fit_transform(X_train).ravel())
```

```python
Xi_train, Xi_test, yi_train, yi_test = train_test_split(Xi, yi, test_size=0.2, random_state=42)

sc = StandardScaler()
Xi_train_s = sc.fit_transform(Xi_train)   # fit on train only
Xi_test_s = sc.transform(Xi_test)         # transform only

clf = LogisticRegression(max_iter=200)
clf.fit(Xi_train_s, yi_train)
print("accuracy:", clf.score(Xi_test_s, yi_test))
```

```python
good = StandardScaler().fit(X_train)      # correct: learn from train
bad = StandardScaler().fit(X_test)        # WRONG: learns from test

print("correct:", good.transform(X_test).ravel())
print("WRONG  :", bad.transform(X_test).ravel())
```
